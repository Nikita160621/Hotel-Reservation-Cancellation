## 1. Imports

```python
import sys; sys.path.append('..')
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
from src.data_loader import load_data, quality_report
from src.eda import run_all_eda
from src.preprocess import prepare
from src.train import baseline_knn, k_sweep, tune_knn
from src.evaluate import compute_metrics, text_report, plot_confusion, plot_k_sweep, plot_roc
```

## 2. Load data and quality check

```python
df = load_data()
display(df.head())
print(quality_report(df))
df.info()
```

## 3. Exploratory data analysis
Figures are saved to `reports/figures/`.

```python
run_all_eda(df)
from IPython.display import Image, display
for f in ['01_target_distribution','02_lead_time','04_cancel_by_segment','08_correlation_heatmap']:
    display(Image(f'../reports/figures/{f}.png'))
```

## 4. Preprocessing, split and scaling

```python
X_train, X_test, y_train, y_test, scaler, features = prepare(df)
print(X_train.shape, X_test.shape)
```

## 5. Baseline KNN (k = 5)

```python
base = baseline_knn(X_train, y_train, k=5)
compute_metrics(base, X_test, y_test)
```

## 6. Choosing k with 5-fold cross-validation

```python
k_values, scores = k_sweep(X_train, y_train)
plot_k_sweep(k_values, scores)
display(Image('../reports/figures/09_k_vs_accuracy.png'))
```

## 7. Hyper-parameter tuning (GridSearchCV)

```python
search = tune_knn(X_train, y_train)
print(search.best_params_)
best = search.best_estimator_
```

## 8. Final evaluation

```python
print(compute_metrics(best, X_test, y_test))
print(text_report(best, X_test, y_test))
plot_confusion(best, X_test, y_test)
plot_roc(best, X_test, y_test)
display(Image('../reports/figures/10_confusion_matrix.png'))
display(Image('../reports/figures/11_roc_curve.png'))
```
